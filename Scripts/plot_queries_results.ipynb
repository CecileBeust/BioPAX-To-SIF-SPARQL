```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
pathbank_meaning1 = pd.read_csv("../Results/PathBank/Meaning1/SIF/PathbankMeaning1.csv", sep=",", header=0)
reactome_meaning1 = pd.read_csv("../Results/ReactomeHomoSapiens95/Meaning1/SIF/ReactomeMeaning1.csv", sep=",", header=0)
print(reactome_meaning1.head())
```

```python
pathbank_meaning1 = pathbank_meaning1[pathbank_meaning1["SPARQL queries"] != "TOTAL"]
reactome_meaning1 = reactome_meaning1[reactome_meaning1["SPARQL queries"] != "TOTAL"]
```

```python
plt.figure(figsize=(12, 10))

plt.scatter(
    x=reactome_meaning1["SPARQL queries"],
    y=reactome_meaning1["Number of nodes"],
    s=reactome_meaning1["Number of edges"]/10,
    alpha=0.7,
    label="Reactome",
    color="blue"
)

plt.scatter(
    x=pathbank_meaning1["SPARQL queries"],
    y=pathbank_meaning1["Number of nodes"],
    s=pathbank_meaning1["Number of edges"]/10,
    alpha=0.7,
    label="PathBank",
    color="orange"
)

plt.title("Meaning 1 SPARQL queries applied to Reactome and PathBank - Size = number of edges")
plt.xlabel("Meaning1 SPARQL queries")
plt.ylabel("Number of nodes")
plt.xticks(rotation=45, ha='right')
plt.grid(True, alpha=0.3)

plt.legend(scatterpoints=1, markerscale=0.1, frameon=True)
plt.tight_layout()
plt.show()
```
